```python
def sum(a,b):
    c=a+b
    return (c)
def diff(a,b):
    c=a-b
    return (c)
def prod(a,b):
    c=a*b
    return (c)
def div(a,b):
    c=a/b
    return (c)
def rem(a,b):
    c=a%b
    return (c)
x=float(input("Enter the first number: "))
y=float(input("Enter the second number: "))
print("Enter(+) if you want to add,(-)if to subtract,(*)if to multiply,(/)if to divide,(%)if to remainder.")
z=input("Enter the operation you want to perform: ")
if z=="+":
    print("The sum is: ",sum(x,y))
elif z=="-":
    print("The difference is: ",diff(x,y))
elif z=="*":
    print("The product is: ",prod(x,y))
elif z=="/":
    if y!=0:
        print("The quotient is: ",div(x,y))
    else:
         print("Cannot divide by zero.")
elif z=="%":
    print("The remainder is: ",rem(x,y))
else:
    print("Invalid choice")


```
